```python
# A strip plot is used to visualize the distribution of numerical values across different categorical groups. It displays individual data points and helps us compare groups.
# Without jitter, many dots may overlap, making it difficult to identify individual observations.

# With jitter=True, points spread slightly, making the distribution easier to understand. 

# Uses random horizontal jitter to reduce overlapping.

# Some dots may still overlap.
```

Box Plot (Box-and-Whisker Plot):

1. Minimum (Lower whisker)

The bottom line represents the lowest non-outlier value.

2. Q1 (First Quartile)

The bottom edge of the box represents 25% of the data.

3. Median (Middle line)

The line inside the box represents the median (50%).

4. Q3 (Third Quartile)

The top edge of the box represents 75% of the data.

5. Maximum (Upper whisker)

The top line represents the highest non-outlier value.
4. How to interpret YOUR graph?
Let's compare the four groups.

A. Others – Male 
- Median is approximately 67.
- 50% of students' marks are between approximately 61 and 77.
- The remaining students are outside this range.
B. Others – Female 
- Median is approximately 74.
- 50% of students' marks are between approximately 65 and 78.
- There is one visible outlier around 41.
C. Central – Male 
- Median is approximately 65.
- 50% of students' marks are between approximately 60 and 73.
D. Central – Female 
- Median is approximately 67.
- 50% of students' marks are between approximately 60 and 75.
